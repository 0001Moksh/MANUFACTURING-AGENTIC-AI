```python
import urllib.parse
import certifi
from pymongo import MongoClient

# 1. Credentials Setup
username = "nofackai_db_user"
raw_password = "mXjvNVuQBKReaMns"  # Replace with your actual password

# Escape special characters in password automatically
password = urllib.parse.quote_plus(raw_password)

# 2. MongoDB Atlas Connection String
uri = (
    f"mongodb+srv://{username}:{password}"
    "@cluster0.aiz2qf6.mongodb.net/"
    "?retryWrites=true&w=majority&appName=Cluster0"
)

try:
    # 3. Initialize MongoClient
    client = MongoClient(
        uri,
        tlsCAFile=certifi.where()
    )

    # 4. Verify connection
    client.admin.command("ping")
    print("Successfully connected to MongoDB Atlas!")

    # 5. Database & Collection
    db = client["my_database"]
    collection = db["licenses"]

    # 6. Sample license records
    sample_records = [
        {
            "customer_name": "iiot_devs",
            "license_token": "IIOT-DEV-8888-XXXX",
            "status": True
        },
        {
            "customer_name": "client_01",
            "license_token": "CLIENT-01-9999-XXXX",
            "status": True
        }
    ]

    # 7. Insert records
    result = collection.insert_many(sample_records)

    print(f"Inserted {len(result.inserted_ids)} license records.")

    # 8. Display inserted records
    for record in collection.find(
        {"_id": {"$in": result.inserted_ids}},
        {"_id": 0}
    ):
        print(record)

except Exception as e:
    print(f"Connection failed: {e}")

finally:
    try:
        client.close()
    except:
        pass
```
